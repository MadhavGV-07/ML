Program 1: Variables and Data Types Write a Python program to declare variables of type int, float, string, and boolean, and display their values along with their data types. Write a program to check the data type of different variables using the type() function. Modify the values of the variables and observe how their data types change. Create your own variables representing a student's details and print their values and data types.

```python
# Program 1: Variables and Data Types

# 1. Declare variables of different data types
num = 25
price = 99.99
name = "Vaishnavi"
is_student = True

print("----- Initial Variables -----")
print("Integer Value:", num)
print("Data Type:", type(num))

print("Float Value:", price)
print("Data Type:", type(price))

print("String Value:", name)
print("Data Type:", type(name))

print("Boolean Value:", is_student)
print("Data Type:", type(is_student))

# 2. Check data types using type()
print("\n----- Checking Data Types -----")
print(type(num))
print(type(price))
print(type(name))
print(type(is_student))

# 3. Modify variable values and observe data type changes
print("\n----- Modifying Variable -----")

x = 100
print("Value:", x)
print("Type:", type(x))

x = 45.67
print("Value:", x)
print("Type:", type(x))

x = "Python Programming"
print("Value:", x)
print("Type:", type(x))

x = False
print("Value:", x)
print("Type:", type(x))

# 4. Student Details
print("\n----- Student Details -----")

student_name = "Vaishnavi"
student_age = 18
student_percentage = 92.5
student_branch = "CSE"
is_hosteller = False

print("Student Name:", student_name)
print("Data Type:", type(student_name))

print("Student Age:", student_age)
print("Data Type:", type(student_age))

print("Student Percentage:", student_percentage)
print("Data Type:", type(student_percentage))

print("Student Branch:", student_branch)
print("Data Type:", type(student_branch))

print("Hosteller:", is_hosteller)
print("Data Type:", type(is_hosteller))
```

Program 2: Operators Write a Python program to demonstrate arithmetic operators. Write a program to demonstrate comparison operators using two numbers. Write a program to demonstrate logical operators (and, or, not). Write a program to determine whether a student is eligible for placement based on CGPA and attendance using logical operators.

```python
# Program 2: Operators

# -------------------------------
# 1. Arithmetic Operators
# -------------------------------

a = 20
b = 10

print("----- Arithmetic Operators -----")
print("a =", a)
print("b =", b)

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Exponentiation:", a ** b)

# -------------------------------
# 2. Comparison Operators
# -------------------------------

print("\n----- Comparison Operators -----")

print("a == b :", a == b)
print("a != b :", a != b)
print("a > b  :", a > b)
print("a < b  :", a < b)
print("a >= b :", a >= b)
print("a <= b :", a <= b)

# -------------------------------
# 3. Logical Operators
# -------------------------------

x = True
y = False

print("\n----- Logical Operators -----")

print("x and y =", x and y)
print("x or y =", x or y)
print("not x =", not x)
print("not y =", not y)

# -------------------------------
# 4. Placement Eligibility
# -------------------------------

cgpa = 8.2
attendance = 85

print("\n----- Placement Eligibility -----")
print("CGPA =", cgpa)
print("Attendance =", attendance)

eligible = (cgpa >= 7.5) and (attendance >= 75)

print("Eligible for Placement:", eligible)
```

Program 4: Lists and Indexing
Write a Python program to create a list and access its elements using indexing.
Write a program to display the first, second, and last elements of a list.
Write a program to demonstrate list slicing.
Write a program to add a new element to an existing list using the append() method.
Write a program to display the total number of elements in a list.

```python
# Program 4: Lists and Indexing

# Create a list
fruits = ["Apple", "Banana", "Mango", "Orange", "Grapes"]

print("----- Original List -----")
print(fruits)

# ---------------------------------
# 1. Access elements using indexing
# ---------------------------------

print("\n----- Accessing Elements -----")
print("First Element:", fruits[0])
print("Second Element:", fruits[1])
print("Third Element:", fruits[2])
print("Last Element:", fruits[-1])

# ---------------------------------
# 2. Display first, second and last
# ---------------------------------

print("\n----- First, Second and Last Elements -----")
print("First:", fruits[0])
print("Second:", fruits[1])
print("Last:", fruits[-1])

# -------------------------
# 3. Demonstrate List Slicing
# -------------------------

print("\n----- List Slicing -----")
print("First Three Elements:", fruits[0:3])
print("Last Two Elements:", fruits[-2:])
print("Elements from Index 1 to 3:", fruits[1:4])
print("Complete List:", fruits[:])

# ---------------------------------
# 4. Add a new element using append()
# ---------------------------------

fruits.append("Pineapple")

print("\n----- After append() -----")
print(fruits)

# ---------------------------------
# 5. Display total number of elements
# ---------------------------------

print("\n----- Total Number of Elements -----")
print("Length of List:", len(fruits))
```

Program 5: Dictionaries
Write a Python program to create a dictionary and display all its key-value pairs.
Write a program to access the value associated with a specific key in a dictionary.
Write a program to iterate through a dictionary using the items() method.
Write a program to store different machine learning models in a dictionary and display their names and objects.

```python
# Program 5: Dictionaries

# -----------------------------------
# 1. Create a dictionary and display all key-value pairs
# -----------------------------------

student = {
    "Roll No": 101,
    "Name": "Vaishnavi",
    "Branch": "CSE",
    "CGPA": 9.2
}

print("----- Student Dictionary -----")
print(student)

print("\nKey-Value Pairs:")
for key, value in student.items():
    print(key, ":", value)

# -----------------------------------
# 2. Access the value using a specific key
# -----------------------------------

print("\n----- Accessing Specific Values -----")
print("Student Name:", student["Name"])
print("Student CGPA:", student["CGPA"])

# -----------------------------------
# 3. Iterate through the dictionary using items()
# -----------------------------------

print("\n----- Iterating using items() -----")
for key, value in student.items():
    print(key, "->", value)

# -----------------------------------
# 4. Store Machine Learning Models in a Dictionary
# -----------------------------------

ml_models = {
    "Linear Regression": "Predicts continuous values",
    "Logistic Regression": "Used for classification",
    "Decision Tree": "Tree-based prediction model",
    "Random Forest": "Ensemble of decision trees",
    "Support Vector Machine": "Classification algorithm",
    "K-Nearest Neighbors": "Instance-based learning"
}

print("\n----- Machine Learning Models -----")
for model, description in ml_models.items():
    print(model, ":", description)
```

Program 6: Conditional Statements
Write a Python program to classify a student's CGPA into High, Mid, or Low categories using if-elif-else.
Write a program to check whether missing values are present in a dataset using conditional statements.
Write a program to determine whether a student is eligible for placement based on CGPA.
Write a program that uses if-elif-else to assign grades based on marks.

```python
# Program 6: Conditional Statements

# ---------------------------------------
# 1. Classify Student CGPA
# ---------------------------------------

cgpa = float(input("Enter Student CGPA: "))

print("\n----- CGPA Classification -----")

if cgpa >= 8.5:
    print("Category: High")
elif cgpa >= 6.5:
    print("Category: Mid")
else:
    print("Category: Low")

# ---------------------------------------
# 2. Check Missing Values in a Dataset
# ---------------------------------------

dataset = [45, 67, None, 89, 92]

print("\n----- Checking Missing Values -----")

if None in dataset:
    print("Missing values are present in the dataset.")
else:
    print("No missing values in the dataset.")

# ---------------------------------------
# 3. Placement Eligibility
# ---------------------------------------

cgpa = float(input("\nEnter CGPA for Placement: "))

print("\n----- Placement Eligibility -----")

if cgpa >= 7.5:
    print("Student is Eligible for Placement.")
else:
    print("Student is Not Eligible for Placement.")

# ---------------------------------------
# 4. Assign Grades Based on Marks
# ---------------------------------------

marks = int(input("\nEnter Student Marks: "))

print("\n----- Grade Calculation -----")

if marks >= 90:
    print("Grade: A+")
elif marks >= 80:
    print("Grade: A")
elif marks >= 70:
    print("Grade: B")
elif marks >= 60:
    print("Grade: C")
elif marks >= 50:
    print("Grade: D")
else:
    print("Grade: F")
```
